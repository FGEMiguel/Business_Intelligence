```python
import pandas as pd
df = pd.read_csv("data\AtencionesUrgencia2024.csv", sep=";", encoding="latin1") 
print(df.head())
print(df.groupby("Idcausa")["TOTAL"].sum().sort_values(ascending=False).head(10))
df.groupby("fecha")["TOTAL"].sum().plot()
```

```python
tramos = ["Menores_1", "De_1_a_4", "De_5_a_14", "De_15_a_64", "De_65_y_mas"]

# 1. Totales nacionales por tramo etario
print(df[tramos].sum())

# 2. % de atenciones de cada tramo etario que ocurre en establecimientos NO hospitalarios
no_hosp = df[df["GLOSATIPOESTABLECIMIENTO"] != "Hospital"]
print("\n% del tramo atendido fuera de Hospital:")
for t in tramos:
    total = df[t].sum()
    fuera = no_hosp[t].sum()
    print(f"{t}: {fuera/total*100:.1f}%")

# 3. Top 5 causas nacionales
causas = df[~df["GlosaCausa"].str.contains("TOTAL|SECCIÓN", na=False)]
print("\nTop 5 causas por Total:")
print(causas.groupby("GlosaCausa")["Total"].sum().sort_values(ascending=False).head(5))
```

```python
import pandas as pd

# tramos etarios en columnas separadas -> a formato largo para sumar por tipo de establecimiento
tramos = ["Menores_1", "De_1_a_4", "De_5_a_14", "De_15_a_64", "De_65_y_mas"]

# 1. Totales nacionales por tramo etario
print(df[tramos].sum())

# 2. % de atenciones de cada tramo etario que ocurre en establecimientos NO hospitalarios
no_hosp = df[df["GLOSATIPOESTABLECIMIENTO"] != "Hospital"]
print("\n% del tramo atendido fuera de Hospital:")
for t in tramos:
    total = df[t].sum()
    fuera = no_hosp[t].sum()
    print(f"{t}: {fuera/total*100:.1f}%")

# 3. Top 5 causas nacionales (excluyendo filas resumen tipo 'TOTAL DEMÁS CAUSAS')
causas = df[~df["GlosaCausa"].str.contains("TOTAL|SECCIÓN", na=False)]
print("\nTop 5 causas por Total:")
print(causas.groupby("GlosaCausa")["Total"].sum().sort_values(ascending=False).head(5))

# 4. guarda un gráfico simple
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
shares = [no_hosp[t].sum()/df[t].sum()*100 for t in tramos]
plt.figure(figsize=(6,4))
plt.bar(tramos, shares)
plt.ylabel("% atendido fuera de Hospital")
plt.title("Share of emergency attentions outside hospital, by age group (2024)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("age_facility_share.png", dpi=150)
print("\ngráfico guardado en age_facility_share.png")
```
